# What is AWS?

**Amazon Web Services (AWS)** is a comprehensive and widely adopted cloud platform offered by Amazon. It provides on-demand cloud computing resources and services like compute power, storage, databases, networking, analytics, machine learning, and more. AWS helps organizations reduce the need for physical infrastructure and scale resources as needed.

---

# Key AWS Services and Their Relevance to Data Science & Machine Learning

Here’s an overview of important AWS services, especially those useful in data science and machine learning:

---

## 1. Compute Services

- **EC2 (Elastic Compute Cloud):** Virtual machines to run applications.  
  _Use Case in DS/ML:_ Run Jupyter notebooks, train ML models on custom environments (e.g., PyTorch, TensorFlow).

- **AWS Lambda:** Serverless computing for running code in response to events.  
  _Use Case in DS/ML:_ Automate data preprocessing when a file is uploaded to S3.

---

## 2. Storage Services

- **S3 (Simple Storage Service):** Scalable object storage.  
  _Use Case in DS/ML:_ Store datasets, model outputs, logs, and serialized models (`.pkl`, `.h5` files).

- **EBS (Elastic Block Store):** Block storage for EC2.  
  _Use Case in DS/ML:_ High-speed storage attached to VMs for faster I/O operations.

---

## 3. Database Services

- **RDS (Relational Database Service):** Managed SQL databases like MySQL, PostgreSQL.

- **DynamoDB:** Managed NoSQL database.  
  _Use Case in DS/ML:_ Store preprocessed data or metadata about models.

---

## 4. Analytics Services

- **Athena:** Query S3-stored data using SQL.

- **Glue:** Managed ETL (Extract, Transform, Load) service.  
  _Use Case in DS/ML:_ Clean and transform raw datasets before model training.

---

## 5. Machine Learning Services

- **SageMaker:** A fully managed service for building, training, and deploying ML models.  
  _Use Case in DS/ML:_ End-to-end ML workflows — from model building (via notebooks) to training and deployment via APIs.

- **Comprehend:** NLP service for sentiment analysis, entity recognition.

- **Rekognition:** Image and video analysis.  
  _Use Case in DS/ML:_ Use prebuilt models for specific tasks (e.g., customer sentiment analysis or face detection).

---

## 6. Orchestration and Automation

- **Step Functions:** Coordinate multiple AWS services into workflows.

- **CloudWatch:** Monitoring service.  
  _Use Case in DS/ML:_ Monitor ML pipelines and automate retraining when performance drops.

---

# Example DS/ML Workflow on AWS

Let’s consider a **predictive maintenance project**:

1. **Data Ingestion:** Sensor data is uploaded to **S3**.
2. **ETL:** **AWS Glue** cleans and transforms the data.
3. **Exploration & Modeling:** Use **SageMaker** notebooks to analyze and train a model.
4. **Model Deployment:** Deploy model as an API using a **SageMaker endpoint**.
5. **Monitoring:** **CloudWatch** tracks performance, and **Step Functions** retrain the model monthly.
